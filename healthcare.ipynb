```python
df_clean["Age"].describe()
```

The dataset contains 54,966 patient records, with ages ranging from 13 to 89 years. The average age is approximately 51.5 years and the median age is 52 years. The middle 50% of patients are approximately between 35 and 68 years old.

```python
import matplotlib.pyplot as plt
plt.hist(df_clean["Age"],bins = 10 )
plt.xlabel("Age")
plt.ylabel("Number of patients")
plt.title("Age distrubtion of patients")
plt.show()
```

Patient ages range from 13 to 89 years, with records broadly distributed across the adult age groups. The distribution is relatively even across much of the 20–80 age range, while the youngest and oldest age ranges contain fewer records.

```python
import matplotlib.pyplot as plt
df_clean.boxplot (column="Age", by= "Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Age")
plt.title("Age distribution by medical condition")
plt.show()
```

The age distributions across medical conditions are broadly similar. Median ages are around 51–52 years, and the interquartile ranges are also comparable across the groups.

```python
df_clean["Gender"].value_counts()
```

```python
pd.crosstab(df_clean["Gender"], df_clean["Medical Condition"])
```

```python
pd.crosstab(
    df_clean["Gender"],
    df_clean["Medical Condition"],
    normalize="index"
) * 100
```

Within this dataset, the six medical conditions account for roughly 16–17% each of the male and female records. The distributions are therefore quite similar between the two gender groups.

```python
df_clean.groupby("Medical Condition")["Billing Amount"].mean()
```

```python
import matplotlib.pyplot as plt
df_clean.groupby("Medical Condition")["Billing Amount"].mean().plot(kind="bar")

plt.xlabel("Medical Condition")
plt.ylabel("Average Billing Amount")
plt.title("Average Billing Amount by Medical Condition")
plt.xticks(rotation=45)
plt.show()
```

Average billing amounts were relatively similar across all medical conditions, ranging from approximately 25,152 to 25,804.

```python
df_clean["Admission Type"].value_counts()
```

Patient admissions were relatively evenly distributed across elective, urgent, and emergency admissions, with elective admissions being slightly more frequent

```python
df_clean["Medication"].value_counts()
```

Medication usage was relatively evenly distributed, with Lipitor being the most frequently recorded medication and Penicillin the least frequent.

```python
df_clean["Medication"].value_counts().plot(kind="bar")

plt.xlabel("Medication")
plt.ylabel("Number of Patients")
plt.title("Medication Distribution")
plt.xticks(rotation=45)
plt.show()
```

```python
df_clean["Test Results"].value_counts()
```

Test results were relatively evenly distributed across abnormal, normal, and inconclusive categories, with abnormal results being slightly more frequent.

```python
df_clean["Test Results"].value_counts().plot(kind="bar")

plt.xlabel("Test Results")
plt.ylabel("Number of Patients")
plt.title("Test Results Distribution")
plt.xticks(rotation=45)
plt.show()
```

```python
pd.crosstab(
    df_clean["Admission Type"],
    df_clean["Medical Condition"]
)
```

Admission types were relatively evenly distributed across the six medical conditions, with no major imbalance in the counts.

```python
pd.crosstab(
    df_clean["Admission Type"],
    df_clean["Medical Condition"],
    normalize="columns"
) * 100
```

```python
gender_condition = pd.crosstab(
    df_clean["Gender"],
    df_clean["Medical Condition"],
    normalize="index"
) * 100

gender_condition.plot(kind="bar")

plt.xlabel("Gender")
plt.ylabel("Percentage of Patients")
plt.title("Medical Condition Distribution by Gender")
plt.xticks(rotation=0)
plt.legend(title="Medical Condition")
plt.show()
```

```python
test_condition = pd.crosstab(
    df_clean["Test Results"],
    df_clean["Medical Condition"]
)

test_condition
```

```python
test_condition.plot(kind="bar", stacked=True)

plt.xlabel("Test Results")
plt.ylabel("Number of Patients")
plt.title("Test Results Distribution by Medical Condition")
plt.xticks(rotation=45)
plt.legend(title="Medical Condition")
plt.show()
```

Test results are relatively evenly distributed across the six medical conditions, with no major imbalance in the number of Normal, Abnormal, and Inconclusive results.
